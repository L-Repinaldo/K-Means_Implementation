```python
from src.core import generate_data, plot_data
from src.k_means import KMeans
```

```python
data=generate_data.generate_data(500)

plot_data.plot(data, title="Generated Data")
```

```python

k = 2

kmeans = KMeans(n_clusters=k)
kmeans.fit(data)

plot_data.plot(
    data,
    centroids=kmeans.centroids_,
    labels=kmeans.labels_,
    title=f"K-Means Clustering - {k} clusters"
)
```
